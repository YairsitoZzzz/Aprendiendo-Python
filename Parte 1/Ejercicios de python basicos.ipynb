```python
# Solución que debe escribir el estudiante
expresion = "5 * 3"

# Usar la estructura match para evaluar la expresión
def evaluar_expresion(expr):
    # Dividir la expresión en sus componentes
    componentes = expr.split()
    
    # Evaluar la expresión usando match
    match componentes:
        case [str(a), "+", str(b)]:
            return int(a) + int(b)
        case [str(a), "-", str(b)]:
            return int(a) - int(b)
        case [str(a), "*", str(b)]:
            return int(a) * int(b)
        case [str(a), "/", str(b)]:
            if int(b) == 0:
                return "Error: División por cero"
            return int(a) / int(b)
        case _:
            return "Expresión no válida"

resultado = evaluar_expresion(expresion)

# Imprimir el resultado
print(f"El resultado de '{expresion}' es: {resultado}")
```

```python
#Solucion que debe escribir el estudiante
def es_primo(num):
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

# Definir el número a verificar
numero = 1

# Utilizar la función es_primo para determinar si el número es primo
resultado = es_primo(numero)

# Imprimir el resultado
print(f"¿El número {numero} es primo? {resultado}")
```

```python
def factorial(n):
    resultado = 1
    while n > 1:
        resultado *= n
        n -= 1
    return resultado

# Definir el número para calcular el factorial
n = 5

# Calcular el factorial utilizando la función definida
resultado = factorial(n)

# Imprimir el resultado
print(f"El factorial de {n} es {resultado}")
```

```python
#Solución que debe escribir el estudiante
def secuencia_fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    
    secuencia = [0, 1]
    i = 2
    while i < n:
        siguiente = secuencia[-1] + secuencia[-2]
        secuencia.append(siguiente)
        i += 1
    return secuencia[:n]

# Probar la función
n = 10
fibonacci = secuencia_fibonacci(n)
print(f"La secuencia de Fibonacci hasta {n} términos es {fibonacci}")
```

```python
# Solución que debe escribir el estudiante
# Definir la lista de números del 1 al 20
numeros = list(range(1, 21))

# Recorrer la lista de números
resultados = []
for num in numeros:
    # Si el número es 15, romper el bucle
    if num == 15:
        break
    # Si el número es par, continuar con la siguiente iteración
    if num % 2 == 0:
        continue
    # Agregar el número a la lista de resultados
    resultados.append(num)

# Imprimir los resultados
print(resultados)
```
